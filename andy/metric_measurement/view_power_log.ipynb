```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# plt.style.use("dark_background")

file_name = 'replica_2000_online.csv'
power_df = pd.read_csv(f'/home/campus.ncl.ac.uk/c4071391/Projects/SLAM3R/data/measurements/{file_name}')
power_df.head(5)

x = power_df['Time (s)'].values
gpu_power = power_df['GPU Power (W)']
cpu_power = power_df['CPU Power (W)']
average_power = power_df['Average Power (W)']
gpu_mem = power_df['GPU Memory Used (GB)']

fig, ax1 = plt.subplots()
ax1.plot(x, gpu_power, label='gpu', color='blue')
ax1.plot(x, cpu_power, label='cpu', color='green')
ax1.plot(x, average_power, label='average', color='orange')
ax2 = ax1.twinx()
ax2.plot(x, gpu_mem, label= 'gpu', color='red')
ax1.set_ylabel('Power [W]')
ax2.set_ylabel('Memory [GB]')
ax1.set_xlabel('Time [s]')

# match axis colours to labels
ax1.tick_params(axis='y', colors='blue')
ax1.yaxis.label.set_color('blue')

ax2.tick_params(axis='y', colors='red')
ax2.yaxis.label.set_color('red')

leg1 = ax1.legend(title='Power', loc='upper left')
leg1.get_frame().set_edgecolor('blue')
leg2 = ax2.legend(title='Memory', loc='upper right')
leg2.get_frame().set_edgecolor('red')
ax1.add_artist(leg1)
ax1.set_title(f'Measurements for file: {file_name}')

plt.show()
```
